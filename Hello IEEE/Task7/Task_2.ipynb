```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid (x):
    return 1/(1 + np.exp(-x))
```

this is the sigmoid function implementaion using np exp so we dont need to loop manually using slow ass python scripts

```python
x_values = np.linspace(-10, 10, 400)
y_values = sigmoid(x_values)

plt.figure(figsize=(8, 6))

plt.plot(x_values, y_values, label=r'Sigmoid: $\sigma(x) = \frac{1}{1 + e^{-x}}$', color='blue', linewidth=2)

plt.axhline(y=1, color='red', linestyle='--', alpha=0.6, label='upper Asymptote (y=1)')
plt.axhline(y=0, color='red', linestyle='--', alpha=0.6, label='lower Asymptote (y=0)')


plt.axhline(y=0.5, color='gray', linestyle=':', alpha=0.6, label='Midpoint (y=0.5)')
plt.axvline(x=0, color='black', linewidth=1)

plt.title('Sigmoid Activation Function', fontsize=16, fontweight='bold')
plt.xlabel('Input (x)', fontsize=12)
plt.ylabel('Output (sigma(x))', fontsize=12)
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.legend(loc='lower right', fontsize=10)

plt.tight_layout()
plt.show()
```

The horizontal lines at y = 1 and y = 0 visually prove that the function squashes all real-numbered inputs into a strictly bounded probability range between 0 and 1
